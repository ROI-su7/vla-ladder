## 每一课的结构

每课都按同一个套路组织，养成固定的学习节奏：

1. **为什么要学**：这个知识点在 VLA / 机器人里用在哪。
2. **原理**：从直觉讲到公式，公式里每个符号都会解释。
3. **代码**：先看一个完整的小例子，逐行注释。
4. **练习**：一个填空函数（带 `TODO`）。
   - 下面紧跟一个折叠起来的 **💡 参考答案**：
     - 在网站 / GitHub 上：点一下就展开。
     - 在 Colab 上：答案单元的代码默认隐藏，点击单元左侧的"显示代码"（或双击单元）即可查看。
     - 在 JupyterLab 上：答案单元默认折叠，点左侧的蓝色竖条展开。
   - 再下面是 **✅ 自测** 单元：运行它，看到"通过"就说明你写对了。
5. **小结与思考题**。

### 关于"参考答案单元"的一个重要约定

**运行答案单元会用参考实现覆盖你写的同名函数。** 后面的代码往往依赖前面练习的函数，如果你实在写不出来，就展开答案、读懂、运行它，然后继续往下走，不要卡死在一个地方。但请务必遵守下面的"三遍法"：

1. **第一遍**：独立写，允许查文档（PyTorch 官方文档、本项目前面的课程），不看答案。
2. **第二遍**：卡住超过 20 分钟，展开答案，**只看思路不复制**，关掉答案自己写出来。
3. **第三遍**：第二天，把这个函数在一个空白文件里从零再写一遍。

第三遍写出来了，才说明你真正掌握了。
